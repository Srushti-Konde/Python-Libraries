```python
#web scraping

import requests

URL = "https://www.scrapethissite.com/pages/"
res = requests.get(URL)

print(res.status_code)
```

```python
if res.status_code == 200:
    # print(res.text)
    # print(res.content)
    print(res.headers) 

with open("scraped_data.html", "w") as f:
    f.write(res.text)
```

```python
from bs4 import BeautifulSoup

with open("scraped_data.html" , "r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
```

```python
soup.find("h1")
soup.h3
soup.find_all("h3")

all_h3 = soup.find_all("h3")
type(all_h3)

for h3 in all_h3:
    print(h3.get_text(strip=True))
```

```python
# parent of h3

all_h3 = soup.find_all("h3")
type(all_h3)

for h3 in all_h3:
    print(h3.get_text(strip=True))
    print(h3.find_parent("div")["class"])# to access particular attrinute


population = h3.find_next("div").select("span.country-population").get_text(strip=True())
```
